```python
def get_city_data(city, api_key):
    baseurl = "http://api.weatherapi.com/v1/"
    current_url = f"{baseurl}current.json?key={api_key}&q={city}"
    forecast_url = f"{baseurl}forecast.json?key={api_key}&q={city}&days=3"
    today = pd.Timestamp.now().strftime("%Y-%m-%d")
    astronomy_url = f"{baseurl}astronomy.json?key={api_key}&q={city}&dt={today}"

    # Get current weather data
    current_response = requests.get(current_url)
    if current_response.status_code != 200:
        raise Exception(f"Error fetching current weather data: {current_response.status_code}")
    current_data = current_response.json()['current']

    # Get forecast data
    forecast_response = requests.get(forecast_url)
    if forecast_response.status_code != 200:
        raise Exception(f"Error fetching forecast data: {forecast_response.status_code}")
    forecast_data = forecast_response.json()['forecast']['forecastday']

    # Get astronomy data
    astronomy_response = requests.get(astronomy_url)
    if astronomy_response.status_code != 200:
        raise Exception(f"Error fetching astronomy data: {astronomy_response.status_code}")
    astronomy_data = astronomy_response.json()['astronomy']['astro']
    
    #Get sunrise and sunset times
    sunrise = astronomy_data['sunrise']
    sunset = astronomy_data['sunset']

    # Process forecast summary
    forecast_3_days_summary = []
    for day in forecast_data:
        date_object = datetime.strptime(day['date'], '%Y-%m-%d')
        formatted_date = date_object.strftime('%A, %B %#d')
        forecast_3_days_summary.append(f"{formatted_date}: High {day['day']['maxtemp_f']}F / Low {day['day']['mintemp_f']}F - {day['day']['condition']['text']}")

    return {
        "current": current_data,
        "forecast": forecast_3_days_summary,
        "astronomy": astronomy_data
    }
```
