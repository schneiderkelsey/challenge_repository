## Step 2: Review the BAD Charts

Below are two deliberately poor visualizations.

```python
plt.figure(figsize=(8,5))
subset = coffee_df.groupby("Year")["Sales"].sum().reset_index()

# Wrong variable for title, truncated y-axis, misleading color & framing
plt.plot(subset["Year"], subset["Sales"], color="red", linewidth=4, marker="o")
plt.ylim(subset["Sales"].min() * 0.90, subset["Sales"].min() * 1.02)  # extreme truncation
plt.title("Collapse in Sales Profits", fontsize=14, fontweight="bold", color="black")
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.show()
```

```python
# Your GOOD visualization(s) here
plt.figure(figsize=(10,6))
subset = coffee_df.groupby("Year")["Sales"].sum().reset_index()

# Example data
data = {
    "Year": [2020, 2021, 2022, 2023, 2024],
    "Sales": [2.9, 2.92, 2.94, 2.96, 2.98, 3.00]
}

#Create a Dataframe
plt.plot(subset["Year"], subset["Sales"], color='red', linewidth=4, marker="o")

#Add labels and title
plt.xlabel("Quarterly Sales Data")
plt.ylabel("Sales")
plt.title("Yearly Sales Trend") 

#Show the plot
plt.grid(True) 
plt.show()
```

```python
plt.figure(figsize=(10,6))
avg_sales = coffee_df.groupby("Region")["Sales"].mean().reset_index()

colors = ["red","orange","yellow","green","blue","purple"]

plt.bar(avg_sales["Region"], avg_sales["Sales"], 
        color=colors, edgecolor="black", linewidth=1, hatch="/", alpha=0.8) #changed the size of the linewidth
plt.title("Sales Data by Region", fontsize=18, fontweight="normal", color="black", backgroundcolor="white") #changed the title, fontweight and color
plt.grid(True, which='both', color='black', linestyle='-', linewidth=1.0) #changed color and linewidth of grid
plt.xlabel("Regions", fontsize=16, fontweight="normal", color="black") #changed fontweight and color
plt.ylabel("Profits", fontsize=16, fontweight="normal", color="black") #changed fontweight and color
plt.xticks(rotation=45, fontsize=14, fontweight="normal", color="black") #changed fontweight and color
plt.yticks(fontsize=14, fontweight="normal", color="black") #changed fontweight and color
plt.show()
```

```python
sales_vs_profit = coffee_df.groupby("Region").agg({"Sales":"mean", "Profit":"mean"}).reset_index()
plt.figure(figsize=(10,6))
plt.scatter(sales_vs_profit["Sales"], sales_vs_profit["Profit"], color="blue", s=100, alpha=0.7, edgecolors="black")
plt.title("Average Sales vs. Average Profit by Region", fontsize=18, fontweight="bold")
plt.xlabel("Average Sales", fontsize=14)
plt.ylabel("Average Profit", fontsize=14)
plt.grid(True)
plt.show()
```

```python
sales_vs_profit.plot(x='Region', y=['Sales', 'Profit'], kind='bar', width=0.5, figsize=(13,5), color=['red','green'])
plt.legend()
plt.ylabel('Amount')
plt.title('Total Sales vs Total Profit by Region')
```
